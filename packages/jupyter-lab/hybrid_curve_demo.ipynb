```python
import matplotlib.pyplot as plt
import pandas as pd

################################# USER DEFINED SPACE ##########################
# Set your x_axis and initialise the empty y_axis.
x_axis = [22 * i / 35 for i in range(-2, 34)]
y_axis = []
# Iterate over x_axis.
####################################
p_lo = 1/4
p_hi = 4
L = 10
for x in x_axis:
    # EQUATION IS ENTERED HERE.
    y = L**2 / (x + L / p_hi**0.5) - L * p_lo**0.5
    # Append result to y_axis list.
    y_axis.append(y)
#####################################

def AMM_x(y):
    return L**2 / (y + L * p_lo**0.5) - L / p_hi**0.5
def AMM_y(x):
    return L**2 / (x + L / p_hi**0.5) - L * p_lo**0.5
# Set your x and y markers that appear at the end of the positive x and y axis lines.
xmarker = ''
ymarker = ''
# Change your labels
x_label = ''
y_label = ''
title = ''
# Choose an appropriate step between x and y ticks.
x_step = 1
y_step = 1
###############################################################################
# Custom X and Y ticks.
for i in range(round(max(x_axis))+1,round(max(x_axis))+x_step+1):
    if i % x_step == 0:
        pos_x_max = i
for i in range(round(min(x_axis))-1,round(min(x_axis))-x_step-1,-1):
    if i % x_step == 0:
        neg_x_max = i
for i in range(round(max(y_axis))+1,round(max(y_axis))+y_step+1):
    if i % y_step == 0:
        pos_y_max = i
for i in range(round(min(y_axis))-1,round(min(y_axis))-y_step-1,-1):
    if i % y_step == 0:
        neg_y_max = i
#xticks = [i for i in range(neg_x_max,pos_x_max+1,x_step)]
#yticks = [i for i in range(neg_y_max,pos_y_max+1,y_step)]
xticks = []
yticks = []
# Change figure size/
plt.figure(figsize=(8,8))
#plt.grid(False, 'both', 'both')

# Plot the graph.
# Vertical and horizontal center lines.
plt.axhline(y = 0, color = 'black', linestyle = '-') 
plt.axvline(x = 0, color = 'black', linestyle = '-')
# Pricing Curve
plt.plot(x_axis, y_axis, color="orange", linestyle='--')
# Secant Lines
plt.plot([0, AMM_x(0)], [AMM_y(0), 0], color="green")
plt.plot([0, AMM_x(2)], [AMM_y(0), 2], color="green", alpha=0.5)
plt.plot([0, AMM_x(4)], [AMM_y(0), 4], color="green", alpha=0.5)
plt.plot([0, AMM_x(6)], [AMM_y(0), 6], color="green", alpha=0.5)
plt.plot([0, AMM_x(8)], [AMM_y(0), 8], color="green", alpha=0.5)
plt.plot([0, AMM_x(10)], [AMM_y(0), 10], color="green", alpha=0.5)
p0 = [3, AMM_y(3)]
p1 = [AMM_y(3), 3]
#plt.plot([p0[0], p1[0]], [p0[1], p1[1]], color="green")
#plt.plot([p0[0]], [p0[1]], marker="o", color="green")
#plt.plot([p1[0]], [p1[1]], marker="o", color="green")
# Lo Tangent
plt.plot([AMM_x(0) + 1/p_lo, AMM_x(0) - 1/p_lo], [-1, 1], color="blue")
plt.plot([AMM_x(0)], [0], marker="o", color="blue")
# Hi Tangent
plt.plot([-1, 1], [AMM_y(0) + p_hi, AMM_y(0) - p_hi], color="red")
plt.plot([0], [AMM_y(0)], marker="o", color="red")

# Change the margin.
#plt.margins(x=0.1,y=0.1)
#Alter your x,y labels and title.
plt.xlabel(xlabel=x_label,fontsize=10)
plt.ylabel(ylabel=y_label,fontsize=10)
plt.title(label=title)

# Annotate the end of the vertical and horizontal lines with your designated markers.
#plt.text(
#        # Xaxis
#        xticks[-1]+x_step/10,
#        # Yaxis
#        y_step / -3,
#        xmarker,
#        fontsize=10)
#plt.text(
#        # Xaxis
#        x_step / -10,
#        # Yaxis
#        yticks[-1]+y_step/10,
#        ymarker,
#        fontsize=10)
# Annotate the center lines with the x and y ticks and draw a grid.
#for x in xticks:
#    plt.text(x,x_step/10,x,rotation=0)
#    if x != 0:
#        plt.axvline(x = x, color = 'grey', linestyle = '--')
#for y in yticks:
#    plt.text(x_step/20,y,y,rotation=0)
#    if y != 0:
#        plt.axhline(y = y, color = 'grey', linestyle = '--') 
# Remove default x and y ticks.
plt.xticks([],fontsize=14)
plt.yticks([],fontsize=14)
# Create the table.
table = pd.DataFrame({
    'X Axis':x_axis,
    'Y Axis':y_axis
})
# Print the table.
#print(table)
# Gradient or Slope calculation.
run = x_axis[1] - x_axis[0]
rise = y_axis[1] - y_axis[0]
gradient = rise / run
# print('')
# print('Gradient:',gradient)
plt.savefig('plots/hybrid_curve_demo.pdf', bbox_inches='tight')
```
